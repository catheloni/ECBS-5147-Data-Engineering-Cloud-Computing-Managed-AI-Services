### 1: Scraping daily star and prothom alo

```python
### Import required libraries for web scraping
import requests
from bs4 import BeautifulSoup
```

```python
### request with proper headers
url1 = "https://www.thedailystar.net/news/bangladesh/politics/news/seat-sharing-bnp-still-keeps-allies-hanging-4062181"
url2 = "https://en.prothomalo.com/bangladesh/politics/n8wdif39or"

headers = {
    "User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10_10_1) AppleWebKit/537.36 (KHTML, like Gecko) "
    + " Chrome/39.0.2171.95 Safari/537.36"
}

star_response = requests.get(url1, headers=headers)
prothomalo_response = requests.get(url2, headers = headers)
```

```python
### parse
daily_star = BeautifulSoup(star_response.content, "html.parser")
prothomalo = BeautifulSoup(prothomalo_response.content, "html.parser")
```

```python
### find title and body of daily star article
title_star = daily_star.find("h1").get_text(strip= True)
star_body = daily_star.find("article")
star_paragraphs = star_body.find_all("p")
star_text = "\n".join([p.get_text(strip=True) for p in star_paragraphs])
print(title_star, "\n", star_text[:500])
```

```python
### find title and body of prothomalo article
prothomalo_title = prothomalo.find("h1").get_text(strip= True)
prothomalo_body = prothomalo.find_all("div", class_="story-element-text")
prothomalo_text = "\n".join([p.get_text(strip=True) for p in prothomalo_body])
print(pothomalo_title, "\n", prothomalo_text[:500])
```
